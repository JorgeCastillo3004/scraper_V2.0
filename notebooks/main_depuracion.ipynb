# Navegar en todos los resultados encontrados y completar informacion.

```python

from collections import defaultdict
from scripts.fix_live_matches import (
    get_pending_live_matches, find_results_url,
    load_until_date, scan_results_page, update_match_in_db
)

# 1. Obtener partidos pendientes
matches = get_pending_live_matches()

# 2. Cargar leagues_info
leagues_info = load_json('check_points/leagues_info.json')

# 3. Agrupar por liga
by_league = defaultdict(list)
for m in matches:
    by_league[(m['sport_name'], m['country_name'], m['league_name'])].append(m)

# 4. Procesar cada liga
stats = {'ok': 0, 'not_found': 0, 'error': 0}

for (sport_name, country_name, league_name), league_matches in by_league.items():
    results_url = find_results_url(leagues_info, sport_name, country_name, league_name)
    if not results_url:
        print('[WARN] URL no encontrada: %s_%s' % (country_name, league_name))
        stats['not_found'] += len(league_matches)
        continue

    print('\n%s' % ('─' * 65))
    print('[%s] %s — %s' % (sport_name, country_name, league_name))
    print('URL: %s' % results_url)

    wait_update_page(driver, results_url, 'container__heading')
    dismiss_cookies(driver)

    oldest_date = min(m['match_date'] for m in league_matches)
    load_until_date(driver, oldest_date)

    found = scan_results_page(driver, league_matches)

    for m in league_matches:
        if m['match_id'] not in found:
            stats['not_found'] += 1
            continue
        try:
            ok = update_match_in_db(driver, m, found[m['match_id']], dry_run=False)
            stats['ok' if ok else 'error'] += 1
        except Exception as e:
            print('  [ERROR] %s: %s' % (m['name'], e))
            stats['error'] += 1

print('\n%s' % ('=' * 65))
print('RESUMEN FINAL:')
print('  Actualizados → COMPLETED : %d' % stats['ok'])
print('  No encontrados           : %d' % stats['not_found'])
print('  Errores                  : %d' % stats['error'])
print('=' * 65)
```

# Fix partidos pasados

Cierra partidos con `match_date < hoy` que estén en estado `LIVE` o que tengan algún `score_entity.points = -1` (típicamente `SCHEDULED` con resultado nunca actualizado por la sección live).

Usa el driver activo del notebook — no abre otro Firefox. Permite correr por lotes filtrando por deporte y/o rango de fechas.


```python
from collections import defaultdict
from scripts.fix_live_matches import (
    get_pending_live_matches, find_results_url,
    load_until_date, scan_results_page, update_match_in_db,
)
from common_functions import load_json, wait_update_page, dismiss_cookies


def fix_past_matches(driver, *, only_status=None, sport=None,
                     date_from=None, date_to=None, dry_run=True):
    """
    Cierra partidos pasados (LIVE o con score=-1) usando el driver activo.

    Parámetros:
      only_status : 'LIVE' | 'SCHEDULED' | None (ambos)
      sport       : nombre de sport.name (ej. 'Football') o None
      date_from   : 'YYYY-MM-DD' o None
      date_to     : 'YYYY-MM-DD' o None
      dry_run     : True imprime sin tocar la DB; False actualiza.

    Devuelve dict con conteos { ok, not_found, error, skipped }.
    """
    matches = get_pending_live_matches(
        only_status=only_status, sport=sport,
        date_from=date_from, date_to=date_to,
    )
    if not matches:
        print('\nSin partidos pendientes para los filtros indicados.')
        return {'ok': 0, 'not_found': 0, 'error': 0, 'skipped': 0}

    leagues_info = load_json('check_points/leagues_info.json')
    by_league = defaultdict(list)
    for m in matches:
        by_league[(m['sport_name'], m['country_name'], m['league_name'])].append(m)

    stats = {'ok': 0, 'not_found': 0, 'error': 0, 'skipped': 0}

    for (sport_name, country_name, league_name), league_matches in by_league.items():
        results_url = find_results_url(leagues_info, sport_name, country_name, league_name)

        print('\n%s' % ('─' * 65))
        print('[%s] %s — %s (%d partido/s)' % (
            sport_name, country_name, league_name, len(league_matches)))

        if not results_url:
            print('  [WARN] URL no encontrada en leagues_info.json')
            stats['not_found'] += len(league_matches)
            continue

        try:
            wait_update_page(driver, results_url, 'container__heading')
            dismiss_cookies(driver)
        except Exception as e:
            print('  [ERROR] navegando: %s' % e)
            stats['error'] += len(league_matches)
            continue

        oldest_date = min(m['match_date'] for m in league_matches)
        load_until_date(driver, oldest_date)
        found = scan_results_page(driver, league_matches)

        for m in league_matches:
            if m['match_id'] not in found:
                stats['not_found'] += 1
                continue
            try:
                ok = update_match_in_db(driver, m, found[m['match_id']], dry_run=dry_run)
                stats['ok' if ok else 'skipped'] += 1
            except Exception as e:
                print('  [ERROR] %s: %s' % (m['name'], e))
                stats['error'] += 1

    print('\n%s' % ('=' * 65))
    print('RESUMEN  (dry_run=%s):' % dry_run)
    print('  Actualizados → COMPLETED : %d' % stats['ok'])
    print('  No encontrados           : %d' % stats['not_found'])
    print('  Omitidos                 : %d' % stats['skipped'])
    print('  Errores                  : %d' % stats['error'])
    print('=' * 65)
    return stats


# ─── Ejemplos de uso ─────────────────────────────────────────────────────────
# Dry-run (no escribe nada) — preview de lo que se haría:
# fix_past_matches(driver, dry_run=True)

# Solo SCHEDULED con score=-1 (excluye los LIVE):
# fix_past_matches(driver, only_status='SCHEDULED', dry_run=True)

# Por lote: solo Football, último mes, sin confirmación por partido:
# fix_past_matches(driver, sport='Football',
#                  date_from='2026-04-01', date_to='2026-04-30', dry_run=False)

# Corrida real masiva (cuidado):
# fix_past_matches(driver, dry_run=False)
```

# Fix equipos faltantes en `league_team` / `team`

Complementa al fix anterior. Detecta equipos referenciados por `match.name` que
no estan registrados en `league_team` para esa liga+temporada (causa raiz de
los partidos "parcialmente reparables"). Clasifica cada faltante en:

- **REGISTRATION_MISSING** — existe en `team`, solo falta la fila en `league_team`.
- **AMBIGUOUS** — varios `team_id` posibles con el mismo nombre+sport en el
  mismo pais → requiere revision manual.
- **TEAM_MISSING** — no existe en `team`; con `apply_creates=True` se crea
  usando el `country_id` de la liga.

**Flujo recomendado:**

1. `run_fix_missing_teams(dry_run=True)` → revisar reporte.
2. `run_fix_missing_teams(dry_run=False, apply_creates=True)` → registrar + crear.
3. `run_fix_inconsistent(dry_run=False)` → completar `match_detail`.


```python
from scripts.fix_missing_teams import fix_missing_teams


def run_fix_missing_teams(*, dry_run=True, apply_creates=False,
                          sport=None, league=None, verbose=True):
    """Wrapper para el notebook — llama a la funcion principal del script.

    Args:
        dry_run       : True (default) solo imprime; False inserta en DB.
        apply_creates : Si True (y dry_run=False) crea teams nuevos para los
                        TEAM_MISSING. Si False, solo registra los que ya
                        existen en `team`.
        sport         : filtra por sport.name (ej. 'Football').
        league        : filtra por league.league_name.
        verbose       : detalle por equipo.

    Returns:
        dict { total_missing, registered, created, ambiguous, skipped, errors }
    """
    return fix_missing_teams(
        dry_run=dry_run, apply_creates=apply_creates,
        sport=sport, league=league, verbose=verbose,
    )


# --- Ejemplos de uso -------------------------------------------------------
# 1) Vista previa (no escribe nada):
# run_fix_missing_teams(dry_run=True)

# 2) Solo registrar (no crear teams nuevos):
# run_fix_missing_teams(dry_run=False, apply_creates=False)

# 3) Registrar + crear teams faltantes (recomendado):
# run_fix_missing_teams(dry_run=False, apply_creates=True)

# 4) Limitado a una liga:
# run_fix_missing_teams(dry_run=False, apply_creates=True, league='Liga de Primera')

# Despues de aplicar este fix, corre el otro para completar match_detail:
# run_fix_inconsistent(dry_run=False)
```
